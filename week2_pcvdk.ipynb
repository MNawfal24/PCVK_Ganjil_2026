```python
##TUGAS PRAKTIKUM 1

# Gambar dalam BGR
img_bgr = cv.imread('/content/drive/MyDrive/Pcvdk_Nawfal/Crop1.jpeg')

# Gambar Merah Biru
img_rb = img_bgr.copy()
img_rb[:, :, 1] = 0

# Gambar hijau biru
img_gb = img_bgr.copy()
img_gb[:, :, 2] = 0

# 4. Gabungkan secara horizontal dan tampilkan
final_frame_1 = cv.hconcat((img_rb, img_gb))
print("Kiri: Kanal Merah-Biru | Kanan: Kanal Hijau-Biru")
cv2_imshow(final_frame_1)
```

```python
##TUGAS PRAKTIKUM 2
# melakukan mirroring
flip_vertikal = cv.flip(img_bgr, 0)
flip_horizontal = cv.flip(img_bgr, 1)
flip_keduanya = cv.flip(img_bgr, -1)

final_frame_2 = cv.hconcat((flip_vertikal, flip_horizontal, flip_keduanya))
print("Kiri: Vertikal | Tengah: Horizontal | Kanan: Keduanya")
cv2_imshow(final_frame_2)
```

```python
##TUGAS PRAKTIKUM 3
img_draw = img_bgr.copy()

# Gambar Persegi dan Lingkaran
cv.rectangle(img_draw, pt1=(30, 430), pt2=(270, 760), color=(255, 0, 0), thickness=5)
cv.circle(img_draw, center=(150, 530), radius=80, color=(0, 255, 0), thickness=5)

# Tambah nama
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(img_draw, text='Muhammad Nawfal Mawla Azhar', org=(300, 80),
           fontFace=font, fontScale=1.5, color=(0, 255, 255),
           thickness=3, lineType=cv.LINE_AA)

final_frame_3 = cv.hconcat((img_bgr, img_draw))
print("Kiri: Foto Asli | Kanan: Foto dengan Geometri & Teks")
cv2_imshow(final_frame_3)
```

```python
##TUGAS PRAKTIKUM - APLIKASI

# PEMROSESAN FOTO 1
print("=== Hasil Foto 1 ===")
img1 = cv.imread('/content/drive/MyDrive/Pcvdk_Nawfal/kucing1.jpg')
h1, w1 = img1.shape[:2]

# Perhitugna untuk foto 1
bar_height1 = h1 // 12
font_scale1 = bar_height1 / 35
thickness1 = max(1, int(bar_height1 / 15))

img_label1 = img1.copy()
img_label1[-bar_height1:, :] = 0

org_x1 = w1 // 40
org_y1 = h1 - (bar_height1 // 3)

cv.putText(img_label1, "Dokumentasi Foto", (org_x1, org_y1),
           cv.FONT_HERSHEY_SIMPLEX, font_scale1, (255, 255, 255), thickness1, cv.LINE_AA)

cv2_imshow(img_label1)


# PEMROSESAN FOTO 2
print("Hasil Foto 2")
img2 = cv.imread('/content/drive/MyDrive/Pcvdk_Nawfal/kucing2.jpg')
h2, w2 = img2.shape[:2]

# Perhitungan untuk foto 2
bar_height2 = h2 // 12
font_scale2 = bar_height2 / 35
thickness2 = max(1, int(bar_height2 / 15))

img_label2 = img2.copy()
img_label2[-bar_height2:, :] = 0

org_x2 = w2 // 40
org_y2 = h2 - (bar_height2 // 3)

cv.putText(img_label2, "Dokumentasi Foto", (org_x2, org_y2),
           cv.FONT_HERSHEY_SIMPLEX, font_scale2, (255, 255, 255), thickness2, cv.LINE_AA)

cv2_imshow(img_label2)
```

```python
##TUGAS PRAKTIKUM APLIKASI 2

img_produk = cv.imread('/content/drive/MyDrive/Pcvdk_Nawfal/kucing1.jpg')
h, w = img_produk.shape[:2]

#Mencari patokan terpendek untuk di crop
sisi_pendek = min(h, w)

#menghitung titik tengah
y_center = h // 2
x_center = w // 2
setengah_sisi = sisi_pendek // 2

#crop gambar di tengah
img_crop = img_produk[y_center - setengah_sisi : y_center + setengah_sisi,
                      x_center - setengah_sisi : x_center + setengah_sisi]

#Tambah nama toko
text = "Toko Kucing"
font = cv.FONT_HERSHEY_SIMPLEX
font_scale = sisi_pendek / 300
thickness = max(1, int(sisi_pendek / 300))

#Mengukur dimensi teks
(text_width, text_height), _ = cv.getTextSize(text, font, font_scale, thickness)
text_x = (sisi_pendek - text_width) // 2
text_y = text_height + (sisi_pendek // 20)

cv.putText(img_crop, text, (text_x, text_y), font, font_scale, (0, 0, 255), thickness, cv.LINE_AA)

#Jadikan ukuran 500x500
thumbnail_crop = cv.resize(img_crop, (500, 500))

## HASIL AKHIR

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(img_produk, cv.COLOR_BGR2RGB))
plt.title(f"Foto Asli \n(Ukuran: {w} x {h})")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(thumbnail_crop, cv.COLOR_BGR2RGB))
plt.title("Thumbnail Crop 1:1 \n(Gambar Penuh Tanpa Kanvas)")
plt.axis('off')

plt.show()
```
